# Zjištění parametru modelu k-mean
- vykreslení elbow grafu
- zjištění ideálního počtu clusterů

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
my_arrays = np.load("iris_numpy.npz")
X = my_arrays['arr_0']
X_norm = my_arrays['arr_2']
X_features = my_arrays['arr_3']
```

```python
import joblib
scaler=joblib.load('classification_std_scaler.bin')
encoder=joblib.load('classification_encoder.bin')
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split (X_norm, Y, test_size=0.2)
```

```python
from sklearn.cluster import KMeans
inertia_list = []
for num_clusters in range(1, 10):
    kmeans_model = KMeans(n_clusters=num_clusters, init="k-means++", n_init = 10)
    kmeans_model.fit(X_norm)
    inertia_list.append(kmeans_model.inertia_)
```

```python
plt.figure(figsize=(6,4))
plt.plot(range(1,10),inertia_list)
plt.scatter(range(1,10),inertia_list)
plt.scatter(4, inertia_list[3], marker="X", s=300, c="r")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia Value")
plt.title("Different Inertia Values for Different Number of Clusters")
```

# k-mean model pro zjištěný počet clusterů
- natrénování modelu
- predikce dat

```python
kmean_model = KMeans(n_clusters=4, random_state= 2, n_init=10)
kmean_model.fit(X_train)
```

```python
x_new=scaler.transform ([[1.4, 0.2]])
y_new=kmean_model.predict(x_new)
setosa_k_id=y_new[0]

x_new=scaler.transform ([[4.3, 1.3]])
y_new=kmean_model.predict(x_new)
versicolor_k_id=y_new[0]

x_new=scaler.transform ([[5.1, 2.3]])
y_new=kmean_model.predict(x_new)
virginica_k_id=y_new[0]

x_new=scaler.transform ([[5.1, 2.3]])
y_new=kmean_model.predict(x_new)
my=y_new[0]
map_kmeans=[setosa_k_id, versicolor_k_id, virginica_k_id, my]
print (map_kmeans)
```

```python
kmean_model.cluster_centers_
```

```python
Y_pred_train = kmean_model.predict(X_train)
Y_pred_test = kmean_model.predict(X_test)
```

```python
for i in range(Y_pred_train.shape[0]):
    Y_pred_train[i]=map_kmeans[Y_pred_train[i]]
    
for i in range(Y_pred_test.shape[0]):
    Y_pred_test[i]=map_kmeans[Y_pred_test[i]] 
```

# Vizualizace modelu
- vykreslení obarvených skupin grafů

```python
plt.figure(figsize=(16,6))

plt.subplot(1,2,1)
plt.scatter(X_train[Y_pred_train == map_kmeans[0], 0], X_train[Y_pred_train == map_kmeans[0], 1], s = 50, c = 'purple', label = 'Iris-setosa')
plt.scatter(X_train[Y_pred_train == map_kmeans[1], 0], X_train[Y_pred_train == map_kmeans[1], 1], s = 50, c = 'orange', label = 'Iris-versicolour')
plt.scatter(X_train[Y_pred_train == map_kmeans[2], 0], X_train[Y_pred_train == map_kmeans[2], 1], s = 50, c = 'green', label = 'Iris-virginica')
plt.scatter(X_train[Y_pred_train == map_kmeans[3], 0], X_train[Y_pred_train == map_kmeans[3], 1], s = 50, c = 'blue', label = 'My')
plt.title('Predicted Gender'); plt.xlabel('Annual Income (k$)'); plt.ylabel('Spending Score (1-100)')

plt.scatter(kmean_model.cluster_centers_[:, 0], kmean_model.cluster_centers_[:,1], s = 100, c = 'red', label = 'Centroids')
plt.legend()

plt.subplot(1,2,2)
plt.scatter(X_train[Y_train == map_kmeans[0], 0], X_train[Y_train == map_kmeans[0], 1], s = 50, c = 'purple', label = 'Iris-setosa')
plt.scatter(X_train[Y_train == map_kmeans[1], 0], X_train[Y_train == map_kmeans[1], 1], s = 50, c = 'orange', label = 'Iris-versicolour')
plt.scatter(X_train[Y_train == map_kmeans[2], 0], X_train[Y_train == map_kmeans[2], 1], s = 50, c = 'green', label = 'Iris-virginica')
plt.scatter(X_train[Y_pred_train == map_kmeans[3], 0], X_train[Y_pred_train == map_kmeans[3], 1], s = 50, c = 'blue', label = 'My')
plt.title('True Species'); plt.xlabel('petal_length'); plt.ylabel('petal_width')

plt.legend()
```

```python
plt.figure(figsize=(16,6))

plt.subplot(1,2,1)
plt.scatter(X_test[Y_pred_test == map_kmeans[0], 0], X_test[Y_pred_test == map_kmeans[0], 1], s = 50, c = 'purple', label = 'Iris-setosa')
plt.scatter(X_test[Y_pred_test == map_kmeans[1], 0], X_test[Y_pred_test == map_kmeans[1], 1], s = 50, c = 'orange', label = 'Iris-versicolour')
plt.scatter(X_test[Y_pred_test == map_kmeans[2], 0], X_test[Y_pred_test == map_kmeans[2], 1], s = 50, c = 'green', label = 'Iris-virginica')
plt.scatter(X_train[Y_pred_train == map_kmeans[3], 0], X_train[Y_pred_train == map_kmeans[3], 1], s = 50, c = 'blue', label = 'My')
plt.title('Predicted Gender'); plt.xlabel('Annual Income (k$)'); plt.ylabel('Spending Score (1-100)')

plt.scatter(kmean_model.cluster_centers_[:, 0], kmean_model.cluster_centers_[:,1], s = 100, c = 'red', label = 'Centroids')
plt.legend()

plt.subplot(1,2,2)
plt.scatter(X_test[Y_test == map_kmeans[0], 0], X_test[Y_test == map_kmeans[0], 1], s = 50, c = 'purple', label = 'Iris-setosa')
plt.scatter(X_test[Y_test == map_kmeans[1], 0], X_test[Y_test == map_kmeans[1], 1], s = 50, c = 'orange', label = 'Iris-versicolour')
plt.scatter(X_test[Y_test == map_kmeans[2], 0], X_test[Y_test == map_kmeans[2], 1], s = 50, c = 'green', label = 'Iris-virginica')
plt.scatter(X_train[Y_pred_train == map_kmeans[3], 0], X_train[Y_pred_train == map_kmeans[3], 1], s = 50, c = 'blue', label = 'My')
plt.title('True Species'); plt.xlabel('petal_length'); plt.ylabel('petal_width')
plt.legend()
```

## Výklad modelu
- Popište jednotlivé skupiny
- Na jaké zákazníky se zaměřit?

Skupina s nižším příjmem a nižším utrácením

Zákazníci s omezeným rozpočtem a zároveň nízkou ochotou utrácet.
Pro firmy to bývá skupina s menším přínosem, ale může mít potenciál pro růst loajality např. přes slevy či akční nabídky.
Skupina s nižším příjmem a vyšším utrácením

Zákazníci, kteří i přes nižší příjem utrácejí relativně hodně.
Může jít např. o impulzivní nakupující nebo o zákazníky s menšími úsporami, kteří ale rádi nakupují (někdy tzv. „aspirational“ segment).
Skupina se středním až vyšším příjmem a průměrným utrácením

Zákazníci, kteří mají solidní příjem, ale utrácejí spíše průměrně.
Obvykle se dají motivovat k vyšším útratám např. nabídkou prémiových služeb či věrnostních programů.
Skupina s vyšším příjmem a vyšším utrácením

Pro firmu nejzajímavější zákazníci z hlediska obratu i marže.
Tito lidé mají vyšší kupní sílu, rádi utrácejí a často očekávají kvalitní služby či prémiový sortiment.

Hlavní cílová skupina bývá většinou skupina s vyšším příjmem a vyšším utrácením (protože generuje největší zisky).
Druhotný cíl může být skupina se středním až vyšším příjmem a průměrným utrácením, kde je potenciál je motivovat k větším nákupům (např. přes cílený marketing, věrnostní programy, prémiové nabídky).
Pokud je strategie firmy zaměřená na objem prodeje a dlouhodobou loajalitu, vyplatí se nezapomínat ani na skupinu s nižším příjmem, avšak vyšší mírou utrácení – tam lze nabízet slevy, akce, splátkové programy či jinou formu zvýhodnění, aby se loajalita ještě posílila.
